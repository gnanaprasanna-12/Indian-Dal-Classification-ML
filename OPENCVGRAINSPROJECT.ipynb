```python
row=[]#add converted images
label=[]#add output names

for dal in os.listdir():
    if os.path.isdir(dal):

        for dal_type in os.listdir(dal):

            type_path = os.path.join(dal, dal_type)

            if os.path.isdir(type_path):

                for image in os.listdir(type_path):

                    image_path = os.path.join(type_path, image)

                    ndarray = cv2.imread(image_path, 0)

                    if ndarray is None:
                        print("Skipping:", image_path)
                        continue

                    resized = cv2.resize(ndarray, (28,28))

                    flattened = resized.flatten()

                    row.append(flattened)

                    label.append(dal + "_" + dal_type)

print("Total images:", len(row))
print("Total labels:", len(label))
print("Classes:", set(label))
```

```python
df1=pd.DataFrame(row,dtype=np.uint8)
df1
```

```python
df2=pd.DataFrame(label,columns=["label"])
df2
```

```python
df=pd.concat([df1,df2],axis=1)
df
```

```python
x=df.drop(columns=['label'])
y=df['label']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=23)
```

```python
from sklearn.metrics import accuracy_score
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn=KNeighborsClassifier()
knn
```

```python
knn.fit(x_train,y_train)
knn
```

```python
y_pred=knn.predict(x_test)
y_pred
```

```python
result=pd.DataFrame({
    "Actual":y_test,
    "Predicted":y_pred})
print(result[:21])
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier(random_state=23)
dt.fit(x_train,y_train)
y_pred=dt.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
def prediction(path):
    ndarray=cv2.imread(path,0)
    resized=cv2.resize(ndarray,(28,28))
    flattened_image=resized.flatten()
    return knn.predict([flattened_image])[0]
```
